```python
import re
from graphviz import Digraph

def parse_verilog(verilog_file):
    """
    Parse Verilog code to extract states and transitions.
    """
    states = set()
    transitions = []

    with open(verilog_file, 'r') as file:
        content = file.read()

    # Extract states from typedef enum
    enum_pattern = re.compile(r'typedef\s+enum\s*\{([^}]+)\}\s*state_type;')
    enum_match = enum_pattern.search(content)
    if enum_match:
        states.update([state.strip() for state in enum_match.group(1).split(',')])

    # Extract transitions from always_comb and case statements
    case_pattern = re.compile(r'case\s*\(state_reg\)\s*(.*?)\s*endcase', re.DOTALL)
    transition_pattern = re.compile(r'(\w+)\s*:\s*if\s*\(([^)]+)\)\s*state_next\s*=\s*(\w+);', re.DOTALL)

    case_blocks = case_pattern.findall(content)
    for case_block in case_blocks:
        for match in transition_pattern.finditer(case_block):
            src_state, condition, dst_state = match.groups()
            transitions.append((src_state.strip(), dst_state.strip(), condition.strip()))

    return states, transitions

def generate_fsm(states, transitions, output_file="fsm_diagram"):
    """
    Generate an FSM diagram using Graphviz.
    """
    fsm = Digraph(format='png')
    fsm.attr(rankdir='LR')

    # Add states
    for state in states:
        fsm.node(state)

    # Add transitions
    for src, dst, cond in transitions:
        fsm.edge(src, dst, label=cond)

    fsm.render(output_file)
    print(f"FSM diagram saved to {output_file}.png")

if __name__ == "__main__":
    # Input Verilog file
    verilog_file = "rising_edge.sv"  # Replace with your Verilog file path

    # Parse the Verilog file
    states, transitions = parse_verilog(verilog_file)

    print("Extracted States:", states)
    print("Extracted Transitions:", transitions)

    # Generate the FSM diagram
    generate_fsm(states, transitions)
```

```python
import re
from graphviz import Digraph

def parse_verilog(verilog_file):
    """
    Parse Verilog code to extract states and transitions.
    """
    states = set()
    transitions = []

    with open(verilog_file, 'r') as file:
        content = file.read()

    # Extract states from typedef enum
    enum_pattern = re.compile(r'typedef\s+enum\s*\{([^}]+)\}\s*state_type;')
    enum_match = enum_pattern.search(content)
    if enum_match:
        states.update([state.strip() for state in enum_match.group(1).split(',')])

    # Extract transitions from always_comb and case statements
    case_pattern = re.compile(r'case\s*\(state_reg\)\s*(.*?)\s*endcase', re.DOTALL)
    transition_pattern = re.compile(r'(\w+)\s*:\s*(if\s*\(([^)]+)\)\s*state_next\s*=\s*(\w+);|state_next\s*=\s*(\w+);)', re.DOTALL)

    case_blocks = case_pattern.findall(content)
    for case_block in case_blocks:
        for match in transition_pattern.finditer(case_block):
            src_state = match.group(1).strip() if match.group(1) else "unknown"
            condition = match.group(3).strip() if match.group(3) else "default"
            dst_state = match.group(4).strip() if match.group(4) else (match.group(5).strip() if match.group(5) else "unknown")
            transitions.append((src_state, dst_state, condition))

    return states, transitions

def generate_fsm(states, transitions, output_file="fsm_diagram"):
    """
    Generate an FSM diagram using Graphviz.
    """
    fsm = Digraph(format='png')
    fsm.attr(rankdir='TB')  # Set the diagram layout to vertical
    fsm.node_attr.update(shape='ellipse', style='filled', fillcolor='lightblue', fontname='Helvetica', fontsize='10')

    # Add states
    for state in states:
        fsm.node(state)

    # Add transitions
    for src, dst, cond in transitions:
        fsm.edge(src, dst, label=cond)

    fsm.render(output_file)
    print(f"FSM diagram saved to {output_file}.png")

if __name__ == "__main__":
    # Input Verilog file
    verilog_file = "rising_edge.sv"  # Replace with your Verilog file path

    # Parse the Verilog file
    states, transitions = parse_verilog(verilog_file)

    print("Extracted States:", states)
    print("Extracted Transitions:", transitions)

    # Generate the FSM diagram
    generate_fsm(states, transitions)
```
